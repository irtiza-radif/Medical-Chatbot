```python
from langchain_community.document_loaders import PyPDFLoader, DirectoryLoader  # Loaders to extract text from a single PDF or an entire folder of PDFs
from langchain_text_splitters import RecursiveCharacterTextSplitter          # Splitter to break long text documents into smaller, structural chunks
```

```python
def load_pdf_files(data):
    # Initializes directory loader to scan the specified folder for all files ending in .pdf using the PyPDFLoader engine
    loader = DirectoryLoader(data, glob="*.pdf", loader_cls=PyPDFLoader)
    documents = loader.load()  # Executes the parsing process and extracts the content into an array of document objects
    return documents           # Returns the loaded document data containing pages, text data, and raw file metadata
```

```python
extracted_data = load_pdf_files("data")
```

```python
from typing import List                     # Imports List type hinting to specify list elements in functions
from langchain_core.documents import Document # Imports the standard LangChain structure for tracking text and metadata

def filter_to_minimal_docs(documents: List[Document]) -> List[Document]:
    minimal_docs = []                       # Initializes an empty list to store the stripped-down document objects
    for doc in documents:                   # Loops through every original document item passed into the function
        minimal_doc = Document(             # Creates a fresh, lightweight document object to save system memory
            page_content=doc.page_content,  # Copies over only the raw text content from the original document source
            metadata={"source": doc.metadata.get("source", "")} # Retains only the specific file path string in metadata
        )
        minimal_docs.append(minimal_doc)    # Adds the newly streamlined document object to the tracking array
    return minimal_docs                     # Returns the filtered array containing only text and source path elements
```

```python
minimal_docs = filter_to_minimal_docs(extracted_data)
```

```python
minimal_docs
```

```python
def text_split(minimal_docs):
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=20)
    texts_chunk = text_splitter.split_documents(minimal_docs)
    return texts_chunk
```

```python
texts_chunk = text_split(minimal_docs)
print(f"Number of chunks: {len(texts_chunk)}")
```

```python
from langchain_classic.chains import create_retrieval_chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
```

```python
system_prompt = (

"You are a medical assistant for question-answering tasks."

"Use the following pieces of retrieved context to answer"

"the question. If you don't know the answer, say that you"

"don't know. Use three sentences maximum and keep the"

"answer concise."

"\n\n"

"{context}"

)
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", system_prompt),
        ("human", "{input}")
    ]
)
```

```python
question_answering_chain = create_stuff_documents_chain(chatmodel, prompt)
rag_chain = create_retrieval_chain(retriever, question_answering_chain)
```

```python
response = rag_chain.invoke({"input": "What is Acromegaly and gigantism?"})
print(response["answer"])
```

```python
response = rag_chain.invoke({"input": "What is Acne?"})
print(response["answer"])
```
